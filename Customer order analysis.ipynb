## Task 1: Store Customer Orders

```python
# 1a. List of customer names
customers = ["Alice", "Bob", "Charlie", "Diana", "Ethan",
             "Fiona", "George", "Hannah", "Ian", "Julia"]

# 1b. Order details stored as tuples inside a list: (name, product, price, category)
orders = [
    ("Alice",   "Television", 120.00, "Electronics"),
    ("Bob",     "Blouse",      45.00, "Clothing"),
    ("Charlie", "Blender",     60.00, "Home Essentials"),
    ("Diana",   "Headphones",  85.00, "Electronics"),
    ("Ethan",   "Pants",       40.00, "Clothing"),
    ("Fiona",   "Microwave",  110.00, "Home Essentials"),
    ("George",  "DVD Player",  55.00, "Electronics"),
    ("Hannah",  "Jacket",      95.00, "Clothing"),
    ("Ian",     "Toaster",     30.00, "Home Essentials"),
    ("Julia",   "Monitor",    150.00, "Electronics"),
]

# 1c. Dictionary: key = customer name, value = list of ordered products
customer_orders = {}
for name, product, price, category in orders:
    customer_orders.setdefault(name, []).append((product, price, category))

print("Customers:", customers)
print("Total orders:", len(orders))
print("Customer Orders:")
for name, items in customer_orders.items():
    print(" ", name, "->", items)
```

### Retrieve and modify records using dictionary methods

```python
# Retrieve a record safely
print("Alice's order:", customer_orders.get("Alice"))
print("Unknown customer:", customer_orders.get("Zoe", "No record found"))

# Work on a copy so the main dataset is not modified
demo = dict(customer_orders)

# Add a new customer record
demo.setdefault("Kevin", []).append(("Socks", 10.00, "Clothing"))
# Add a second item for an existing customer
demo["Alice"].append(("Remote", 15.00, "Electronics"))
# Update using .update()
demo.update({"Bob": [("Blouse", 45.00, "Clothing"), ("Scarf", 20.00, "Clothing")]})
# Remove a record
removed = demo.pop("Ian")

print("Modified Alice:", demo["Alice"])
print("Modified Bob  :", demo["Bob"])
print("Added Kevin   :", demo["Kevin"])
print("Removed Ian   :", removed)
print("Original dataset still has", len(customer_orders), "customers")
```

## Task 2: Classify Products by Category

```python
product_categories = {}

# Loop through each order and map the product to its category
for name, product, price, category in orders:
    product_categories[product] = category

# Create a set of all unique categories
print("Product Categories:", product_categories)
unique_categories = set(product_categories.values())
print("Unique Categories:", unique_categories)

# Display all available product categories
print("\nAvailable categories:")
for cat in sorted(unique_categories):
    print(" -", cat)
```

## Task 3: Analyze Customer Orders

```python
# Dictionary to store total spending per customer
customer_totals = {}

# Calculate total amount each customer spends
for customer, items in customer_orders.items():
    total = sum(price for product, price, category in items)
    customer_totals[customer] = total

print("Customer Totals:", customer_totals)

# Classify customers by total spending
customer_class = {}
for customer, total in customer_totals.items():
    if total > 100:
        customer_class[customer] = "High-value buyer"
    elif total >= 50:
        customer_class[customer] = "Moderate buyer"
    else:
        customer_class[customer] = "Low-value buyer"

print("\nCustomer Classification:")
for customer, label in customer_class.items():
    print(f"  {customer:<8} ${customer_totals[customer]:>7.2f}  ->  {label}")
```

## Task 4: Generate Business Insights

```python
# 4a. Total revenue per product category
category_revenue = {}
for name, product, price, category in orders:
    category_revenue[category] = category_revenue.get(category, 0) + price

print("Revenue per category:")
for cat, rev in category_revenue.items():
    print(f"  {cat:<16} ${rev:>7.2f}")

# 4b. Unique products from all orders (set)
unique_products = {product for name, product, price, category in orders}
print("\nUnique products:", unique_products)
print("Number of unique products:", len(unique_products))

# 4c. List comprehension: customers who purchased electronics
electronics_customers = [name for name, product, price, category in orders
                         if category == "Electronics"]
print("\nElectronics customers:", electronics_customers)

# 4d. Top three highest-spending customers using sorting
top_three = sorted(customer_totals.items(), key=lambda x: x[1], reverse=True)[:3]
print("\nTop 3 highest-spending customers:")
for rank, (customer, total) in enumerate(top_three, start=1):
    print(f"  {rank}. {customer:<8} ${total:.2f}")
```

## Task 5: Organize and Display Data

```python
# 5a. Summary of each customer's total spending and classification
print("=" * 48)
print(f"{'Customer':<10}{'Total Spent':>12}   Classification")
print("=" * 48)
for customer in customers:
    print(f"{customer:<10}${customer_totals[customer]:>10.2f}   {customer_class[customer]}")
print("=" * 48)

# 5b. Build a set of customers for each category
category_customers = {}
for name, product, price, category in orders:
    category_customers.setdefault(category, set()).add(name)

for cat, people in category_customers.items():
    print(f"{cat}: {sorted(people)}")

# 5c. Customers who purchased from multiple categories (set operations)
cats = list(category_customers)
multi_category = set()
for i in range(len(cats)):
    for j in range(i + 1, len(cats)):
        multi_category |= category_customers[cats[i]] & category_customers[cats[j]]
print("\nCustomers who bought from multiple categories:", multi_category or "None")

# 5d. Common customers who bought both Electronics and Clothing
common = category_customers["Electronics"] & category_customers["Clothing"]
print("Bought both Electronics and Clothing:", common or "None")

# Extra set operations for comparison
print("Electronics only (difference):", sorted(category_customers["Electronics"] - category_customers["Clothing"]))
print("Electronics or Clothing (union):", sorted(category_customers["Electronics"] | category_customers["Clothing"]))
```

**Summary:** Stored the orders in lists, tuples, dictionaries and sets, then classified customers by spend (3 high, 4 moderate, 3 low). Electronics earned the most ($410 of $790), Julia, Alice and Fiona are the top spenders, and no customer bought across categories.
